```python
import matplotlib.pyplot as plt
import numpy as np
from sklearn.datasets import make_blobs
```

```python
#Generate some blob data
n_samples=300
X,y=make_blobs(n_samples,centers=4,cluster_std=0.60,random_state=42)
```

```python
#Visualize the data

plt.scatter(X[:,0],X[:,1],s=30)
plt.title("Generated Data")
plt.show()
```

```python
from sklearn.cluster import DBSCAN

dbscan=DBSCAN(eps=0.9, min_samples=40)
dbscan.fit(X)
y_dbscan=dbscan.fit_predict(X)
#Visualize the data

plt.scatter(X[:,0],X[:,1],s=30,c=y_dbscan,cmap='viridis')
plt.title("Generated Data")
plt.show()
```

```python
#Compute the k-distance plot

from sklearn.neighbors import NearestNeighbors
neighbors=NearestNeighbors(n_neighbors=10)
neighbors.fit(X)
distances,_ =neighbors.kneighbors(X)

#Sort and plot the distances
distances=np.sort(distances[:,9],axis=0)
plt.plot(distances)
plt.xlabel("Points")
plt.ylabel("Epsilon")
plt.title("K-Distance plot")
plt.show()
```
