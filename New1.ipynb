```python
#Dictionary..............................collection of key value pairs

aDict={
    'eat':'to eat something',
    'act':'To do acting'
}
type(aDict)
aDict['act']

```

```python
bDict={'a':1000, 'b':2000}
print(bDict)
```

```python
#Dictionary methods
#First method ... dict.keys()
print(aDict.keys())
print(type(bDict))
```

```python
for key in bDict.keys():
    print(key)
```

```python
#Dictionary methods
#First method ... dict.items()
print(bDict.items())
```

```python
bDict[1]='c'
print(bDict)
```
