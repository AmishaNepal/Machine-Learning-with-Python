```python
import numpy as np
import matplotlib.pyplot as plt

slope=0.7
intercept=0.3

X=np.arange(0,1,0.02)
y=slope*X+intercept
print(X.shape,y.shape)
```

```python
split=int(0.8*len(X))
X_train,y_train=X[:split],y[:split]
X_test,y_test=X[split:],y[split:]
print("Shape of X_train",X_train.shape)
print("Shape of y_train",y_train.shape)
print("Shape of X_test",X_test.shape)
print("Shape of X_test",X_test.shape)
```

```python
def plot_data(X_train=X_train, y_train=y_train, X_test=X_test, y_test=y_test, predictions=None):
    plt.figure(figsize=(8,6))
    plt.xlabel("X")
    plt.ylabel("y")
    plt.title("X vs y")
    plt.scatter(X_train, y_train,c='b', s=8, label="Training Data")
    plt.scatter(X_test, y_test,c='r', s=8, label="Testing Data")
    if predictions is not None:
        plt.scatter(X_test, predictions,c='g', s=8, label="Predicted Data")
    plt.legend()

plot_data()
```

```python
y_preds=0.5*X_test+0.2
plot_data(predictions=y_preds)
```

```python
X_train_mean = np.mean(X_train)
y_train_mean = np.mean(y_train)
```

```python
m=np.sum((X_train-X_train_mean)*(y_train-y_train_mean))/np.sum((X_train-X_train_mean)**2)

b=y_train_mean-m*X_train_mean

y_preds=m*X_test+b
plot_data(predictions=y_preds)
```

```python
def mse(y_actual, y_preds):
    return np.mean((y_actual-y_preds)**2)
print(f'MSE={mse(y_actual=y_test,y_preds=y_preds):.6f}')
```

```python
print(f"Slope={m:.4f}")
print(f"Intercept={b:.4f}")
```

```python
import numpy as np
import matplotlib.pyplot as plt 
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, mean_absolute_error
slope = 0.7
intercept = 0.3
X = np.arange(0,1,0.02)
y = slope*X + intercept
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=123)
def plot_data(X_train=X_train, y_train=y_train, X_test=X_test, y_test=y_test, predictions=None):
    plt.figure(figsize=(8,6))
    plt.xlabel("X")
    plt.ylabel("y")
    plt.title("X vs y")
    plt.scatter(X_train, y_train,c='b', s=8, label="Training Data")
    plt.scatter(X_test, y_test,c='r', s=8, label="Testing Data")
    if predictions is not None:
        plt.scatter(X_test, predictions,c='g', s=8, label="Predicted Data")
    plt.legend()
plot_data()
```

```python
lr_model = LinearRegression()
lr_model.fit(X_train.reshape(-1, 1), y_train)
y_preds = lr_model.predict(X_test.reshape(-1,1))
plot_data(predictions=y_preds)
```

```python
print(f'MSE = {mean_squared_error(y_true=y_test, y_pred=y_preds):.6f}')
print(f'MAE = {mean_absolute_error(y_true=y_test, y_pred=y_preds):.6f}')
```
