```python
# Create a 1D NumPy array with 10 elements, containing the numbers from 1 to 10.

import numpy as np
array1=np.arange(1,11)
print(array1)
```

```python
# Create a 2D NumPy array of shape (3, 4) filled with ones.

array2=np.arange(1,13)
array2=np.copy(array2.reshape(3,4))
array2[array2 > 1] = 1
print(array2)
```

```python
# Create a 3x3 identity matrix using NumPy.

martrix3=np.arange(20,29).reshape(3,3)
print(martrix3)
```

```python
# Add 5 to each element in a 1D NumPy array: [1, 2, 3, 4, 5].

array3=np.arange(1,6)
print("Original 1D array:\n",array3)
print("Array after adding 5 to each elements:\n",array3 + 5)
```

```python
#Given a 2D array, extract the element from the second row and third column.

arrY=np.arange(1,13)
reshaped=np.copy(arrY.reshape(3,4))
print("The 2D array is:\n",reshaped)
print("The element from second row Third column is:\n",reshaped[1,2])
```

```python
#Find the minimum and maximum value in a NumPy array: arr = np.array([2, 4, 6, 8, 10]).

arrZ = np.arange(2,11)
newarray=arrZ[arrZ % 2 == 0]
print("The array is:\n",newarray)
print("The minimum value is:\n",np.min(newarray))
print("The maximum value is:\n",np.max(newarray))
```

```python
#Calculate the sum and mean of the array: arr = np.array([3, 5, 7, 9]).

arrW = np.arange(3,10)
newarrayA=arrW[arrW % 2 != 0]
print("The array is:\n",newarrayA)
print("The sum value is:\n",np.sum(newarrayA))
print("The mean value is:\n",np.mean(newarrayA))
```

```python
#Reshape a 1D array arr = np.array([1, 2, 3, 4, 5, 6]) into a 2D array of shape (2,3)

ODarray=np.arange(1,7)
print("The 1D array is:\n",ODarray)
TDarray=np.copy(ODarray.reshape(2,3))
print("A 2D array of shape (2,3) is:\n",TDarray)
```

```python
# Flatten a 2D array arr = np.array([[1, 2, 3], [4, 5, 6]]) into a 1D array.

flattened_array = TDarray.flatten()
print(flattened_array)
```

```python
#Stack two 1D arrays [1, 2, 3] and [4, 5, 6] vertically and horizontally.

x=np.array([1,2,3])
y=np.array([4,5,6])

# Stacking these two 1D array vertically
z=np.stack((x,y),axis=1)
print("Vertical stacking:\n",z)

# Stacking these two 1D array horizontally
z=np.stack((x,y),axis=0)
print("Horizontal stacking:\n",z)
```

```python
#Perform element-wise addition and subtraction on two arrays: arr1 = np.array([2, 4, 6]) and arr2 = np.array([1, 3, 5]).

arrayX=np.array([2,4,6])
arrayY=np.array([1,3,5])
print("Element wise addition on two array:\n",arrayX+arrayY)
print("Element wise subtraction on two array:\n",arrayX-arrayY)
```

```python
#Perform element-wise multiplication and division on two arrays: arr1 = np.array([2, 4, 6]) and arr2 = np.array([1, 2, 3]).

arrayX1=np.array([2,4,6])
arrayY1=np.array([1,2,3])
print("Element wise multiplication on two array:\n",arrayX1*arrayY1)
print("Element wise division on two array:\n",arrayX1/arrayY1)
```

```python
#Calculate the dot product of two 1D arrays: arr1 = np.array([1, 2]) and arr2 = np.array([3, 4]).
a=np.array([1,2])
b=np.array([3,4])
print("Dot product of two 1D array:\n",np.dot(a,b))
```

```python
#Multiply two matrices using np.dot(). Given matrices A = np.array([[1, 2], [3, 4]]) and B = np.array([[5, 6], [7, 8]]), compute their dot product.
A=np.array([[1,2],[3,4]])
B=np.array([[5,6],[7,8]])
print("Dot product of two matrices:\n",np.dot(A,B))
```

```python
#Use np.sum(), np.mean(), and np.std() to compute the sum, mean, and standard deviation of the array arr = np.array([10, 20, 30, 40, 50]).

K=np.array([10,20,30,40,50])
print("The sum of elements of the array is:\n",np.sum(K))
print("The mean of elements of the array is:\n",np.mean(K))
print("The Standard Deviation of elements of the array is:\n",np.std(K))
```

```python
#Find the index of the maximum value in a NumPy array arr = np.array([5, 2, 8, 1, 6]) using np.argmax().

L=np.array([5,2,8,1,6])
print("The index of the maximum value is:\n",np.argmax(L))
```
