```python
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt 
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
```

```python
iris = load_iris()
X = iris.data
y = iris.target
feature_names = iris.feature_names

# Convert to DataFrame
df = pd.DataFrame(X, columns=feature_names)
df['Species'] = y
df.head()
```

```python
scaler=StandardScaler()
X_scaled=scaler.fit_transform(df)
```

```python
pca=PCA(n_components=2)
X_2d=pca.fit_transform(X_scaled)
X_2d.shape
```

```python
print("EVR",pca.explained_variance_ratio_)
print("CEV",np.cumsum(pca.explained_variance_ratio_))
```

```python
plt.figure(figsize=(8,6))
plt.scatter(X_2d[:,0],X_2d[:,1],c=y,cmap='viridis',s=100)
plt.show()
```

```python
data_re=pca.inverse_transform(X_2d)
```

```python
X_re_origi=scaler.inverse_transform(data_re)
```
