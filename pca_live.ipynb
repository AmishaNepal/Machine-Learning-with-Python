```python
# Sample data (2 dimensions)
data = np.array([[[2, 3,4],[3, 4,5]],
                [[4, 5,6],[5, 6,7]],
                [[6, 7,8],[7,8,9]]])
print("Shape is",data.shape)
```

```python
#Step 1:Standardize the data

scaler=StandardScaler()
data_standardized=scaler.fit_transform(data)
```

```python
#Step 2:Implement PCA

pca=PCA(n_components=2)
data_pca=pca.fit_transform(data_standardized)
```

```python
from sklearn.datasets import make_classification
```

```python
X,y=make_classification(n_samples=1000,n_features=5,n_informative=3,random_state=42)
```

```python
X[:10]
```

```python
scaler=StandardScaler()
X_scaled=scaler.fit_transform(X)
```

```python
pca=PCA(n_components=2)
X_2d=pca.fit_transform(X_scaled)
```

```python
print("EVR",pca.explained_variance_ratio_)
print("CEV",np.cumsum(pca.explained_variance_ratio_))
```

```python
plt.figure(figsize=(8,6))
plt.scatter(X_2d[:,0],X_2d[:,1],c=y,cmap='viridis',s=100)
plt.show()
```

```python
data_re=pca.inverse_transform(X_2d)
```

```python
X_re_origi=scaler.inverse_transform(data_re)
```
