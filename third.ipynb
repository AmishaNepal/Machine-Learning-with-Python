```python
# Compute the sum of values in aDict using dict.values()
cDict = {'Ram':1000, 'Shyam':2000, 'Hari':3000, 'Gopal':4000}
sum = 0
for value in cDict.values():
    sum = sum + value
print(f'Sum = {sum}')
```

```python
# Compute the sum of values in aDict using dict.keys()
cDict = {'Ram':1000, 'Shyam':2000, 'Hari':3000, 'Gopal':4000}
sum = 0
keyList = cDict.keys()
for key in cDict.keys():
    sum = sum + cDict[key]
print(f'Sum = {sum}')
```

```python
# Compute the sum of values in aDict using dict.items()
cDict = {'Ram':1000, 'Shyam':2000, 'Hari':3000, 'Gopal':4000}
sum = 0
for item in cDict.items():
    sum = sum + item[1]
print(f'Sum = {sum}')
```

```python
# Program to ask for a sentence and count the frequency of letter
sentence = "the quick brown fox jumps over a lazy dog"
countDict = {}
for letter in sentence:
    if letter not in countDict:
        countDict[letter] = 1
    else:
        countDict[letter] = countDict[letter] + 1
print(countDict)
```

```python
from time import time
```

```python
def sum(N):
    s = 0
    for i in range(1,N+1):
        s = s + i
    return s
```

```python
n = int(input("Enter a number "))
start = time()
answer = sum(n)
end = time()
print(f'The sum upto {n} terms is {answer}')
print(f'The time taken is {end-start} seconds')
```

```python
def sum1(N):
    return (N+1)*(N/2)
```

```python
n = int(input("Enter a number "))
start = time()
answer = sum1(n)
end = time()
print(f'The sum upto {n} terms is {answer}')
print(f'The time taken is {end-start} seconds')
```

```python
class Dog:
    def __init__(self, name, breed):
        self.name = name
        self.breed = breed

    def __str__(self):
        return f'Dog Name = {self.name} and Breed = {self.breed}'

    def bark(self):
        print(f'{self.name} says Woof!')

dog1 = Dog('Tommy', 'Tibetian')
dog1.bark()
print(dog1)
```

```python
aString = "RISETOVOTESIR"

def checkPalindrome(aString):
    if aString == aString[::-1]:
        return True
    else:
        return False

print(checkPalindrome(aString))
```

```python
aString = "RISETOVOTESIR"

def checkPalindrome(aString):
    bString = ""
    for i in range(len(aString)-1,-1,-1):
        bString = aString[i] + bString
    if aString == bString:
        return True
    else:
        return False
print(checkPalindrome(aString))
```

```python
# Write a program that keeps asking the user to enter a positive number 
# until they enter a negative number. 
# Once a negative number is entered, the program should stop.
flag = True
while(flag):
    n = int(input("Enter a positive number "))
    if n>=0:
        print(f'You have entered {n}')
    else:
        print(f'You have entered {n} which is a negative number!Bye!!')
        flag = False
```

```python
def countFactor(n):
    c = 0
    for i in range(1,n+1):
        if n%i == 0:
            c = c + 1
    return c
```

```python
def checkPrime(n):
    if(countFactor(n) == 2):
        print(f"{n} is Prime")
    else:
        print(f"{n} is Composite")
checkPrime(11)
```
