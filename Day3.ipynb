```python
#Create two 2*2 matrices with random 
matrix1=np.random.randint(10,50,(2,2))
matrix2=np.random.randint(10,50,(2,2))
print(matrix1)
print(matrix2)
```

```python
print("Addition:\n",matrix1+matrix2)
print("Sub:\n",matrix1-matrix2)
print("Multiply:\n",matrix1*matrix2)
print("Divide:\n",matrix1/matrix2)
```

```python
dot_product=np.dot(matrix1,matrix2)
print("Dot product:\n",dot_product)
```

```python
#row wise mean
print(matrix1.mean(axis=1))
```

```python
#column wise mean
print(matrix1.mean(axis=0))
```
