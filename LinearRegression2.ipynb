```python
import numpy as np
import matplotlib.pyplot as plt 
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, mean_absolute_error

X=pd.read_csv("linear_regression_data.csv")
```

```python
#Creating model
X_train, X_test, y_train, y_test = train_test_split(X['X'], X['Y'], test_size=0.3, random_state=123)
```

```python
def plot_data(X_train=X_train, y_train=y_train, X_test=X_test, y_test=y_test, predictions=None):
    plt.figure(figsize=(8,6))
    plt.xlabel("X")
    plt.ylabel("y")
    plt.title("X vs y")
    plt.scatter(X_train, y_train,c='b', s=8, label="Training Data")
    plt.scatter(X_test, y_test,c='r', s=8, label="Testing Data")
    if predictions is not None:
        plt.scatter(X_test, predictions,c='g', s=8, label="Predicted Data")
    plt.legend()
```

```python
plot_data()
```

```python
lr_model = LinearRegression()
lr_model.fit(X_train, y_train)
y_preds = lr_model.predict(X_test)
plot_data(predictions=y_preds)
```

```python
print(f'MSE = {mean_squared_error(y_true=y_test, y_pred=y_preds):.6f}')
print(f'MAE = {mean_absolute_error(y_true=y_test, y_pred=y_preds):.6f}')
```
