```python
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns 
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
# Generate some synthetic data
X, y = make_blobs(n_samples=20, centers=3, cluster_std=0.8, random_state=42)
print(X[:5], y[:5])
# Visualize the data
plt.scatter(X[:, 0], X[:, 1], s=40)
plt.title("Generated Data")
plt.show()
# Initialize KMeans
kmeans = KMeans(n_clusters=3)
# Fit the model to the data
kmeans.fit(X)
# Predict the cluster
y_kmeans = kmeans.predict(X)
centers = kmeans.cluster_centers_
print(centers)
plt.scatter(X[:, 0], X[:, 1], c=y_kmeans, s=40, cmap='viridis')
plt.scatter(centers[:, 0],centers[:, 1] , s=80, alpha=0.5, marker='x')
plt.title("K-Means Clustering")
plt.show()
# Elbow method
# Calculate the Within-Cluster-Sum-of-Squares(WCSS)
wcss = []
for k in range(1,11):
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(X)
    wcss.append(kmeans.inertia_)

# Plot the Elbow Curve
plt.plot(range(1,11), wcss, marker='o')
plt.xlabel("k")
plt.ylabel("WCSS")
plt.title("Elbow Method")
plt.show()
from sklearn.metrics import silhouette_score
sList = []
for k in range(2,11):
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(X)
    y_kmeans = kmeans.predict(X)
    score = silhouette_score(X, y_kmeans)
    sList.append(score)

plt.plot(range(2,11), sList, marker='o')
plt.xlabel('k')
plt.ylabel("Silhoutte Score")
plt.title('k vs Silhoutte score')
plt.show()
```

```python
n_samples=1000
from sklearn.datasets import make_circles

X,y = make_circles(n_samples,factor=0.5,noise=0.05,random_state=42)
wcss = []
for k in range(1, 21):
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(X)
    y_kmeans = kmeans.predict(X)
    wcss.append(kmeans.inertia_)

plt.plot(range(1,21), wcss, marker='o')
plt.xlabel('k')
plt.ylabel('WCSS')
plt.title('k vs WCSS')
plt.show()
```
